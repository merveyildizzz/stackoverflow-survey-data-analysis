## Question 7: What are the average and median yearly salaries of Python developers across countries?

### Approach

To analyze salary differences among Python users across countries:

- First, we filter respondents who reported using Python.
- Then, we select relevant columns: Country and yearly compensation.
- We convert the salary column to numeric and remove missing values.
- To ensure reliable results, we exclude countries with very small sample sizes.
- Finally, we group the data by country and calculate:
  - Mean (average salary)
  - Median (middle value)

This approach allows us to compare salary distributions across different countries while avoiding misleading results caused by insufficient data.

```python
python_users = df[df['LanguageHaveWorkedWith'].str.contains('Python', na=False)]
```

```python
python_users = python_users[['Country', 'ConvertedCompYearly']]
```

```python
python_users['ConvertedCompYearly'] = pd.to_numeric(
    python_users['ConvertedCompYearly'],
    errors='coerce'
)
```

```python
python_users_clean = python_users.dropna()
```

```python
country_salary = python_users_clean.groupby('Country')['ConvertedCompYearly'].agg(['mean', 'median'])
```

```python
country_salary_sorted = country_salary.sort_values(by='mean', ascending=False)
country_salary_sorted.head(10)
```

```python
country_counts = python_users_clean['Country'].value_counts()
```

```python
valid_countries = country_counts[country_counts > 50].index
```

```python
python_users_filtered = python_users_clean[
    python_users_clean['Country'].isin(valid_countries)
]
```

```python
country_salary = python_users_filtered.groupby('Country')['ConvertedCompYearly'].agg(['mean', 'median'])

country_salary_sorted = country_salary.sort_values(by='mean', ascending=False)

country_salary_sorted.head(10)
```

### Insight

After filtering out countries with small sample sizes, we observe that the highest salaries among Python developers are concentrated in developed countries.

The **United States** leads with the highest average salary (~$173K), followed by Switzerland and Israel, indicating strong compensation in tech-driven economies.

European countries such as Ireland, the United Kingdom, Denmark, and Norway also show consistently high median salaries, suggesting stable and well-distributed income levels.

The differences between mean and median values indicate some level of income inequality, but overall trends remain consistent.

This analysis highlights that geographic location plays a significant role in developer compensation, even within the same technology stack.

## 🎯 Bonus Question 1: What percentage of Python users are there in each age group?

### Approach

To calculate the percentage of Python users in each age group:

- First, we group all respondents by age.
- Then, we identify respondents who reported using Python.
- For each age group, we calculate:
  - Total number of respondents
  - Number of Python users
- Finally, we compute the percentage of Python users within each age group.

This helps us understand how Python usage varies across different age groups.

```python
# Python users
python_users = df[df['LanguageHaveWorkedWith'].str.contains('Python', na=False)]

# Total respondents by age
age_total = df['Age'].value_counts()

# Python users by age
age_python = python_users['Age'].value_counts()

# Combine into one DataFrame
age_analysis = pd.DataFrame({
    'total': age_total,
    'python_users': age_python
}).fillna(0)

# Calculate percentage
age_analysis['python_percentage'] = (age_analysis['python_users'] / age_analysis['total']) * 100

age_analysis['python_percentage'] = age_analysis['python_percentage'].round(2)
# Sort
age_analysis_sorted = age_analysis.sort_values(by='python_percentage', ascending=False)

age_analysis_sorted.head(10)
```

### Insight

Python usage is relatively consistent across most age groups, ranging between approximately 36% and 40%.

The highest adoption rate is observed among the 18–24 age group (~40%), indicating strong early adoption among younger developers.

Interestingly, older age groups (45–64) also show high Python usage (~37–39%), suggesting that Python is widely adopted across experience levels, not only by younger professionals.

The lowest adoption rate appears in the 65+ group (~31%), which may reflect lower engagement with newer technologies.

Overall, this indicates that Python is not limited to a specific age group but is broadly used across the developer population.

## 🎯 Bonus Question 2: What industries are most common among remote workers in the top 25% salary group?

### Approach

To identify the most common industries among high-earning remote workers:

- First, we convert the salary column to numeric format and remove missing values.
- Then, we calculate the 75th percentile (top 25% salary threshold).
- We filter respondents whose salary is above this threshold.
- Next, we identify remote workers by filtering the RemoteWork column (including Remote and Hybrid).
- Finally, we analyze the Industry column to find the most common industries among this group.

This helps us understand which industries dominate among high-income remote professionals.

```python
# Step 1: Prepare salary column
df['ConvertedCompYearly'] = pd.to_numeric(df['ConvertedCompYearly'], errors='coerce')
```

```python
# Step 2: Remove missing salaries
salary_clean = df.dropna(subset=['ConvertedCompYearly'])
```

```python
# Step 3: Calculate top 25% threshold
salary_threshold = salary_clean['ConvertedCompYearly'].quantile(0.75)
```

```python
salary_threshold
```

```python
# Step 4: Filter high earners (top 25%)
high_earners = salary_clean[salary_clean['ConvertedCompYearly'] >= salary_threshold]
```

```python
# Step 5: Filter remote workers (Remote + Hybrid)
remote_workers = high_earners[
    high_earners['RemoteWork'].str.contains('Remote|Hybrid', na=False)
]
```

```python
# Step 6: Analyze industries
industry_counts = remote_workers['Industry'].value_counts()

industry_counts.head(10)
```

```python
# Industry percentage distribution
industry_percentage = (industry_counts / industry_counts.sum()) * 100
industry_percentage.round(2).head(10)
```

## Insight
Software Development clearly dominates both in count and percentage (~43%), making it the leading industry among high-income remote workers.

Other industries such as Fintech, Healthcare, and Telecommunications follow but with significantly lower shares, indicating a strong concentration of high-paying remote roles in tech-driven sectors.

This suggests that remote high-income opportunities are heavily clustered in software and technology-related fields, where digital infrastructure and global collaboration are essential.
